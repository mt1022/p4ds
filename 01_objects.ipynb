## Python程序基本概念的层次结构

### 层次结构
- 程序(Program)由模块组成
- 模块由多条指令(Statement)组成
- 指令包含表达式(Expression)
- 表达式创建、处理对象(Object)
- 对象无处不在 (everything in python is an object)

### 基本概念
- 指令(Statement)：告诉python我们的程序应该做什么的代码，一个指令要么是一个表达式，要么是由某个关键字构成的几种结构之一；
- 表达式(Expression)：用来计算某个值的"对象与操作符的组合"，即表达式就是任何“能算出值来”的代码片段；
- 对象(Object)：数据以对象的形式存在；

```python
# example statement
a = 1 + 1
#   - object
#       - object
#   ----- expression
```

## 变量
变量（variable）就是对象的名字。
- 变量指代某个对象，并不需要提前创建；
- 变量在首次赋值时创建；（即将变量名与具体的某个对象用赋值操作符`=`绑定）
- 变量只有在被赋值之后才能在表达式中使用；
- 变量在表达式中使用时会被替换成对应的对象；

C语言
```C
int a;  // 声明整型变量 a
a = 3;  // 给 a 赋值为 3
```

python
```python
a = 3
```

如下面的示意图所示，a是变量，3是对象，a指代3，这个指代关系也被叫做引用(reference):
```mermaid
flowchart LR
    a --> 3
```
为了方便，在不引起误解的情况下，我们有时候对变量名及其所指代的对象并不做区分，大家需要注意。

````python
a = "python"
````

````python
print(a)
````

````python
"a = {a}"
````

````python
a
````

````python
# f-string会把大括号括起来的表达式求值后嵌入字符串
# 因此下面表达第一个a是字符；第二个a是变量，最终将被替换成其所指代的字符串。
f"a = {a}"
````

#### 常用运算符（operator） (按照优先级从高到低排列)

| 优先级 | 运算符                  | 描述                          | 示例                      | 结合性     |
|--------|-------------------------|-----------------------------|--------------------------|------------|
| 1      | `**`                    | 指数（幂运算）                | `x ** y` (x 的 y 次方)    | 从右到左   |
| 2      | `~`, `+`, `-`           | 按位取反、正号、负号           | `~x`, `+x`, `-x`         | 从右到左   |
| 3      | `*`, `/`, `//`, `%`     | 乘、除、整除、取模             | `x * y`, `x / y`, `x // y`, `x % y` | 从左到右   |
| 4      | `+`, `-`                | 加、减                        | `x + y`, `x - y`         | 从左到右   |
| 5      | `<<`, `>>`              | 左移、右移                    | `x << y`, `x >> y`       | 从左到右   |
| 6      | `&`                     | 按位与                        | `x & y`                  | 从左到右   |
| 7      | `^`                     | 按位异或                      | `x ^ y`                  | 从左到右   |
| 8      | `\|`                    | 按位或                        | `x \| y`                 | 从左到右   |
| 9      | `==`, `!=`, `>`, `<`, `>=`, `<=` | 比较运算符 | `x == y`, `x != y`, `x > y` 等 | 从左到右   |
| 10     | `is`, `is not`, `in`, `not in` | 身份/成员运算符 | `x is y`, `x in y` 等    | 从左到右   |
| 11     | `not`                   | 逻辑非                        | `not x`                  | 从右到左   |
| 12     | `and`                   | 逻辑与                        | `x and y`                | 从左到右   |
| 13     | `or`                    | 逻辑或                        | `x or y`                 | 从左到右   |
| 14     | `:=`                    | 海象运算符 (Python 3.8+)      | `if (n := len(x)) > 10:` | 从右到左   |

注意：结合性决定了相同优先级运算符的计算顺序（如 `2 ** 3 ** 2` 按从右到左计算为 `2 ** (3 ** 2)`）

````python
2 * pi * 3**2
````

````python
(2 * (pi * 3))**2
````

````python
2 * pi * (3**2)
````
