## 对象的属性(attribute)与方法(method)

### 属性（attribute）即对象有什么？

属性是对象"自带的数据"，通过 `object.attribute` 访问（不加括号）。python 中的对象有很多特性，特性可以是普通的对象，即属性(property)，也可以是适用于该对象的函数,即方法（method）。

```python
object.__attribute__
```
前后都有两个下划线的特性，是特殊的特性，一般由 python 代码解释器调用，而非用户直接调用。

例如这些方法在冒号后面对应的指令出现时，由代码解释器隐式调用：
- `x.__iter__`: `iter(x)`
- `x.__str__`: `print(x)`
- `x.__len__`: `len(x)`
- `x.__contains__`: `i in x`
- `x.__getitem__(slice(start, stop, step))`: `x[start:stop:step]`


### 方法（method）即对象能干什么？

| 写法 | 含义 |
|---|---|
| `function(arguments)` | 调用普通函数 |
| `object.method(arguments)` | 调用对象的方法 |

普通函数是独立的，方法则是跟特定对象绑定的函数：

```python
# 普通函数
len("hello")          # 5

# 方法：属于字符串对象
"hello".upper()       # HELLO
"hello".replace("h", "H")  # Hello

# 方法：属于列表对象
names = ["alice", "bob"]
names.append("tom")   # names → ["alice", "bob", "tom"]
```

### 流程控制

#### 开始之前

- **Python 会自动识别指令或代码块的边界**，不需要手动标记结束。
- **一行末尾就是一条指令的结束**，不需要像 C/Java 那样用 `;` 结尾：

```python
x = 10

```

- **可以用 `;` 把几条简单指令写在同一行**（不推荐，影响可读性）：

```python
x = 1; y = 2; print(x + y)

```

  注意：`;` 不能用于把 `if`、`for`、`def`、`class` 等复合语句挤到同一行。

- **指令跨多行时，用括号 `()`、`[]`、`{}` 括起来即可自动续行**：

```python
total = (1 + 2
         + 3 + 4)

```

- **行末加反斜杠 `\` 也能续行**（不推荐，反斜杠后必须紧跟换行符，多一个空格就会报错）：

```python
total = 1 + 2 + \
        3 + 4

```

- **缩进表示代码块的开始，缩进结束即代码块结束**（Python 用缩进代替 `{}`）：

```python
if x > 0:
    print("正数")   # 缩进内 → 属于 if
print("结束")       # 缩进结束 → if 块结束

```

- 缩进的形式可以用tab、四个空格、两个空格等，但在同一个模块中要保持一致，不要混合使用。PEP 8 规范推荐**用 4 个空格**。

##### 反思and、or、not

- python中的任何对象都有一个内在的`True`或`False`值；
- 数字0、空对象、`None`的布尔值是`False`，其他对象的布尔值是`True`；
- 比较(>, <, >=, <=)或相等（==）测试返回的值是`True`/`False`；
- `or` : 如果左侧对象的布尔值是`True`，返回左边的对象，否则返回右边的对象；
- `and`: 如果左侧对象的布尔值是`False`，返回左边的对象，否则返回右边的对象；
- `not`: 返回与右侧对象的布尔值相反的布尔值；

（简单的记忆方法：算到哪里就返回哪里的值，没必要算的部分值不会返回）

````python
x = 2 or 3
print(x)
````

````python
x = 0 or 2
x
````

### Generator (生成器）与 Iterator（迭代器）的区别
**迭代器**是实现了 `__iter__()` 和 `__next__()` 方法的对象，可以逐个取出元素，取完后抛出 `StopIteration`。`for` 循环的本质就是不断调用 `next()`。

**生成器**是一种特殊的迭代器——用带 `yield` 的函数或生成器表达式写成，不需要手写类。

关系：**所有生成器都是迭代器，反之不然**。


````python
import sys

lst = [x for x in range(100000)]
print(sys.getsizeof(lst))

# 生成器表达式 （generator expression)
gen = (x for x in range(100000))
print(sys.getsizeof(gen))
````

#### `nonlocal` 声明非局部变量

```python
def outer():
    x = 0
    def inner():
        nonlocal x
        x += 1
    inner()
    return x

print(outer())
```
